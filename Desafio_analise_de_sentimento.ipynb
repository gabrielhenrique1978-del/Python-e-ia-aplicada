# Análise de Sentimento de Avaliações de Produtos com Groq API

Este notebook demonstra como realizar análise de sentimento em avaliações de produtos utilizando a API da Groq para processamento de linguagem natural. O objetivo é classificar cada avaliação como 'positiva', 'negativa' ou 'neutra' e adicionar essa classificação ao DataFrame original.

## Etapa 1: Carregamento dos Dados

```python
import pandas as pd
# etapa 1
df_reviews = pd.read_csv('/content/reviews (1).csv')
```

Primeiro, importamos a biblioteca `pandas` para manipulação de dados e carregamos o arquivo CSV `reviews (1).csv` em um DataFrame chamado `df_reviews`. Este DataFrame contém as avaliações dos produtos.

```python
df_reviews
```

Exibimos o DataFrame carregado para ter uma visão geral dos dados, incluindo as colunas disponíveis e as primeiras linhas das avaliações.

```python
coluna_review = df_reviews['reviewText']
print(coluna_review)
```

Extraímos a coluna `reviewText` do DataFrame `df_reviews`, que contém o texto das avaliações que serão analisadas. Esta coluna é armazenada em uma variável separada para facilitar o processamento.

## Etapa 2: Configuração da Groq API

Instalamos a biblioteca `groq` para interagir com a API da Groq. O `-q` garante uma instalação silenciosa.

```python
import os
from google.colab import userdata
os.environ['groq_api_key'] = userdata.get('groq_key')
```

Configuramos a chave da API Groq. É uma boa prática armazenar chaves de API em segredos (secrets) do Colab para evitar expor a chave diretamente no código. Aqui, a chave é carregada do ambiente de usuário (`userdata`).

```python
import os
from groq import Groq

client = Groq(api_key=os.environ['groq_api_key'])
completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": "Explain the importance of low-latency LLMs",
        }
    ],
    model="openai/gpt-oss-20b"
)
```

Inicializamos o cliente Groq com a chave da API e fazemos uma chamada de exemplo para garantir que a API está funcionando corretamente. Isso verifica a conectividade e a autenticação com a Groq API.

```python
lista_de_analise_sentimentos =[]

for review_numero, resenha in enumerate(coluna_review):
  resposta = client.chat.completions.create(
      model="openai/gpt-oss-20b",
      messages=[
          {
              "role": "user",
              "content": f'''voce ira analisar a resenha que eu te mandarei abaixo, e retornara com uma analise de sentimento. voce deve responder com uma das seguintes palavras: ''positiva'', ''negativa'' ou ''neutra'', indicando o sentimento relativo a resenha especifica.\n\n              Resenha: {resenha}'''
          }
      ]
  )

  lista_de_analise_sentimentos.append(resposta.choices[0].message.content)
```

## Etapa 3: Análise de Sentimento

Iteramos sobre cada resenha na `coluna_review`, enviando o texto da resenha para a Groq API para análise de sentimento. A API é instruída a retornar o sentimento como 'positiva', 'negativa' ou 'neutra'. Os resultados são armazenados em `lista_de_analise_sentimentos`.

```python
lista_de_analise_sentimentos
```

Exibimos a lista completa de análises de sentimento que foram geradas pela Groq API.

```python
df_reviews['analises de sentimentos'] = lista_de_analise_sentimentos
```

## Etapa 4: Integração dos Resultados

Adicionamos a `lista_de_analise_sentimentos` como uma nova coluna chamada `'analises de sentimentos'` ao DataFrame original `df_reviews`. Isso integra os resultados da análise de sentimento diretamente com os dados das avaliações.
